```python
import itertools

def generate_prompts(
    age_increment: int = 5,
    use_gender: bool = True,
    use_ethnicity: bool = True,
    use_emotion: bool = True,
    output_file: str = "generated_prompts.txt"
):
    """
    Generates a list of prompts based on specified attributes and saves them to a file.

    Args:
        age_increment (int): The step to use for ages (e.g., 5 means 0, 5, 10...).
        use_gender (bool): Whether to include gender in the prompts.
        use_ethnicity (bool): Whether to include ethnicity in the prompts.
        use_emotion (bool): Whether to include emotion in the prompts.
        output_file (str): The name of the file to save the prompts to.
    """
    # --- Define all possible attribute values ---
    ages = list(range(0, 90, age_increment))
    genders = ["male", "female"]
    ethnicities = ["caucasian", "afroamerican", "asian"]
    emotions = {
        'neutral': "with a neutral expression",
        'happy': "smiling happily",
        'slightlyhappy': "smiling slightly",
        'other': "showing a subtle emotion"
    }

    # Create lists of attributes to combine, using a placeholder for disabled ones
    age_list = ages
    gender_list = genders if use_gender else [None]
    ethnicity_list = ethnicities if use_ethnicity else [None]
    emotion_list = list(emotions.keys()) if use_emotion else [None]

    all_prompts = []
    
    # Generate all combinations
    all_combinations = itertools.product(age_list, gender_list, ethnicity_list, emotion_list)

    print("--- Generating Prompts ---")
    
    with open(output_file, "w") as f:
        for age, gender, ethnicity, emotion_key in all_combinations:
            # Dynamically build the prompt parts
            prompt_parts = [f"A {age} years old"]
            
            if ethnicity:
                prompt_parts.append(ethnicity)
            if gender:
                prompt_parts.append(gender)
            if emotion_key:
                prompt_parts.append(emotions[emotion_key])

            prompt = " ".join(prompt_parts)
            
            all_prompts.append(prompt)
            print(prompt) # Print to console
            f.write(prompt + "\n") # Write to file

    print("\n" + "="*30)
    print(f"✅ Generation Complete!")
    print(f"Total prompts generated: {len(all_prompts)}")
    print(f"Saved to: {output_file}")
    print("="*30)

    return all_prompts
```

```python
# Generate prompts with an age increment of 5 and all attributes enabled
generate_prompts(
    age_increment=5,
    use_gender=True,
    use_ethnicity=True,
    use_emotion=True,
    output_file="prompts_age_inc_5.txt"
)

# Example of generating with ethnicity disabled
# generate_prompts(
#     age_increment=10,
#     use_gender=True,
#     use_ethnicity=False,
#     use_emotion=True,
#     output_file="prompts_no_ethnicity.txt"
# )
```
